### LAÇO WHILE


```python
# Usando o loop while para imprimir os valores de 0 a 9
counter = 0
while counter < 10:
    print(counter)
    counter = counter + 1
```

```python
# Também é possível usar a claúsula else para encerrar o loop while
x = 0

while x < 10:
    print ('O valor de x nesta iteração é: ', x)
    print (' x ainda é menor que 10, somando 1 a x')
    x += 1

else:
    print ('Loop concluído!')
```

### Pass, Break, Continue

```python
counter = 0
while counter < 100:
    if counter == 10:
        break
    else:
        pass
    print(counter)
    counter = counter + 1
```

```python
for verificador in "Python":
    if verificador == "h":
        continue
    print(verificador)
```

### While e For juntos

```python
for i in range(2,30):
    j = 2
    counter = 0
    while j < i:
        if i % j == 0:
            counter = 1
            j = j + 1
        else:
            j = j + 1

    if counter == 0:
        print(str(i) + " é um número primo")
        counter = 0
    else:
        counter = 0
```

# Vamos Treinar Juntos

Antes vamos ver mais dois conceitos importantes

Um **acumulador** é uma variável que ocorre em ambos os lados de uma atribuição e que, antes de ser usada pela primeira vez, é iniciada com um valor específico. Por exemplo, efetuando a atribuição a = 3, iniciamos a variável a com o valor 3; então, quando a atribuição a = a + 2 é executada, essa variável passa a ter o valor 5. Note que, se o valor inicial de a não é definido pela primeira atribuição, não é possível determinar o seu valor após a execução da segunda atribuição.

Um **contador** é um tipo de acumulador cujo valor aumenta, ou diminui, de 1 em 1.

```python
spositivos = 0
qnegativos = 0

for i in range(1, 21):
  numero = int(input(f"digite o {i} numero inteiro"))

  if numero > 0:
    spositivos += numero
  elif numero < 0:
    qnegativos += 1

print("/nresultados")
print(f"a soma dos numeros positivos: {spositivos}")
print(f"quantidade de valores negativos: {qnegativos}")
```

6) SOMAR NÚMEROS ATÉ QUE O USUÁRIO DIGITE 0

```python
soma = 0

numero = int(input("Digite um número (0 para parar): "))

while numero != 0:
    soma = soma + numero
    numero = int(input("Digite outro numero (0 para parar): "))

print("A soma dos numeros é:", soma)
```

8) Faça um programa que mostre os n termos da Série a seguir:

image.png

```python
n = int(input("Digite o número de termos da série: "))

soma = 0
denominador = 1

print("Série:")
for numerador in range(1, n + 1):
    termo = numerador / denominador
    print(f"{numerador}/{denominador}", end=" ")
    if numerador < n:
        print("+", end=" ")
    soma += termo
    denominador += 2

print(f"\n\nSoma dos {n} termos da série: {soma:.4f}")
```
