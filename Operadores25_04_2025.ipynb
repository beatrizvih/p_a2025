1-Par ou impar


```python
# Entrada
numero = int(input("Digite um numero: "))

# Processamento
paridade = ["Par", "Ímpar"][numero % 2]

# Saída
print(f"O número {numero} é {paridade}.")
```

2- valor de lados do triangulo

```python
#entrada
l1 = int(input("digite o valor do primeiro lado:"))
l2 = int(input("digite o valor do segundo lado:"))
l3 = int(input("digite o valor do terceiro lado:"))
#processamento
forma_triangulo = (l1 + l2 > l3) and (l1 + l3 > l2) and (l3 + l2 > l1)

equilatero = forma_triangulo and (l1 == l2 == l3)
escaleno = forma_triangulo and ((l1 == l2) or (l1 == l3) or (l2 == l3))
isosceles = forma_triangulo and not equilatero and not isosceles
#saida
print(f"forma um triangulo? {forma_triangulo}")
print(f"todos os lados são iguais? {escaleno}")
print(f"todos os lados diferentes? {isosceles}")
print(f"dois lados iguais? {equilatero}")
```

#imc e peso


```python
# Entrada de dados
peso = float(input("Digite seu peso (kg): "))
altura = float(input("Digite sua altura (m): "))
# Processamento
imc = peso / (altura ** 2)

faixas = [
           (imc < 18.5, "abaixo do peso"),
           (imc < 25, "peso normal"),
           (imc < 30, "sobrepeso"),
           (imc < 35, "obesidade grau 1"),
           (imc < 40, "obesidade grau 2"),
           (imc < 45, "obesidade grau 3"),
]
status = next(texto for cond, texto in faixas if cond)

# Saída
print(f"seu imc é: {imc :.2f}")
print(f"classificação: {status}")
```
