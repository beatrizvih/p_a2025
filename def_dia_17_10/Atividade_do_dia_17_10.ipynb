## 2- conversor em temperatura

crie uma função chamada Celsius_to_fahrenheit que recebe uma temperatura em Celcius(C) e retorna a temperatura equivalente a Fahernheit(F)
**Formula**
(0 °C × 9/5) + 32 = 32 °F

💡Dica lembre-se de se converter a fração para decimal para garantir o cálculo correto  e usar o return para devolver o valor.


```python
def Celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 1.8) + 32
    return fahrenheit
```

```python
print(Celsius_to_fahrenheit(0))
```

## 4 - Quadrado e Cubo

Crie uma função camada calcular_potencias que recebe um número n e retorna o quadrado e o cubo desse número em uma tupla.

💡 A funçao deve retornar dois valores. Lembre que em Python, voce pode retornar multiplos valores, que são automaticamente empacotados em uma tupla.

```python
def calcular_potencias(n):
    quadrado = n ** 2
    cubo = n ** 3
    return (quadrado, cubo)
```

```python
resultado = calcular_potencias(3)
print(resultado)
```

## 5 - Contagem Regressiva

Crie uma função *contagem_regressiva* que recebe um numero inteiro *inicio* até 1

💡 Use um laço de repetição for ou while para iterar e imprimir os numeros sequencialmente. Não é nessesario retornar nada.

```python
def contagem_regressiva(inicio):
    for i in range(inicio, 0, -1):
        print(i)
```

```python
contagem_regressiva(12)
```
