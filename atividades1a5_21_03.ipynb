Atividade numero 3

```python
##entrada
hora = int(input('valor de horas'))
minuto = int(input('valor de minutos'))
valorhoras = float(input('valor horas trabalhada'))
##processamento
conversão = hora+(minuto/60)
salario = conversão*valorhoras
resultado = salario+(salario*70/100)
##saida

print(f"o valor do seu salario com hora extra, feito em base em {conversão} horas é : {resultado}")
```

Lista extra 4

```python
##entrada
capital = int(input('digite o valor da capital:'))
taxa = int(input('digite o valor da taxa:'))
tempo = int (input('digite o valor do tempo:'))
##processamento
taxa = taxa/100
montante = (capital) * (1+taxa)**tempo
##saida
print(f"o resultado é: {montante}")
```

Divisão de Conta em um Restaurante com Gorjeta 5

```python
##entrada
pessoa = int(input('a quantidade de pessoas:'))
gorjeta = int(input('a quantidade de gorjeta:'))
conta = int(input('o valor da conta:'))
##processamento
gorjeta = (conta * 10/100)
conta = (gorjeta + conta)
valorpessoa = (conta + gorjeta)/pessoa
##saida
print (f"o resulatdo é: {valorpessoa}")
```
