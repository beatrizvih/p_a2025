#Como calcular a média

```python
# entrada
nota1 = float(input('A nota da primeira prova é :'))
nota2 = float(input('A nota da segunda prova é :'))

#processamento
media = (nota1 * nota2 ) / 2

#saida
print( 'A média é:', media)


```

#Converter kilometro para milha

```python
#entrada
kilometro = float(input('kilometro:'))
#processamento
milha = 1.609 * kilometro
#saida
print('milhas:', milha)
```
