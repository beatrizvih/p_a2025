```python
dia = int(input("Digite um dia: "))

if dia < 1 or dia > 31:
    print("Data inválida!")
else:
    if dia <= 7:
        print("Lua Nova")
    else:
        if dia <= 14:
           print("Quarto Crescente")
        else:
            if dia <= 21:
                print("Lua Cheia")
            else:
                if dia <= 29:
                    print("Quarto Minguante")
                else:
                    print("Início de novo ciclo lunar")
                    
```
