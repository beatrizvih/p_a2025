##8) Faça um programa que mostre os n termos da Série a seguir e realize o cálculo
Solicite ao usuário a quantidade de repetições

s = (1/1)+(2/3)+(3/5)+(4/7)+...+...n/m

```python
n = int(input("Digite o número de termos da série: "))

soma = 0
denominador = 1

print("Série:")
for numerador in range(1, n + 1):
    termo = numerador / denominador
    print(f"{numerador}/{denominador}", end=" ")
    if numerador < n:
        print("+", end=" ")
    soma += termo
    denominador += 2

print(f"\n\nSoma dos {n} termos da série: {soma:.4f}")
```

##5) Remover elementos repetidos de uma lista
Usando um for, percorra a lista valores = [1, 2, 2, 3, 4, 4, 5] e crie uma nova lista apenas com elementos únicos.

```python
valores = [1, 2, 2, 3, 4, 4, 5]
unicos = []

for valor in valores:
    if valor not in unicos:
        unicos.append(valor)

print("Lista sem repetidos:", unicos)
```

##6) Verificar se palavra é palíndromo
Peça uma palavra ao usuário e use while ou for para verificar se ela é igual de frente para trás (ex.: "arara").

```python
palavra = input("Digite uma palavra: ")
palavra = palavra.lower()  # Deixa tudo minúsculo para evitar erro com maiúsculas

inversa = ""

for i in range(len(palavra) - 1, -1, -1):
    inversa += palavra[i]

if palavra == inversa:
    print("É um palíndromo!")
else:
    print("Não é um palíndromo.")
```

##10) Contador de palavras em uma frase
Peça ao usuário uma frase, divida em palavras (split()) e use um dicionário para contar quantas vezes cada palavra aparece.
Com manipulação de listas, tuplas e dicionários:


```python
frase = input("Digite uma frase: ")

palavras = frase.split()  # Divide a frase em palavras usando espaços
contador = {}  # Dicionário para contar as ocorrências

for palavra in palavras:
    palavra = palavra.lower()  # Deixa tudo minúsculo para evitar duplicatas por maiúsculas
    if palavra in contador:
        contador[palavra] += 1
    else:
        contador[palavra] = 1

# Exibindo os resultados com tuplas
print("\nContagem de palavras:")
for item in contador.items():
    print(item)  # Cada item é uma tupla (palavra, quantidade)
```

##2) Cadastro de Pessoas e Análise de Alturas
Solicite ao usuário para inserir o nome e a altura de várias pessoas. Preencha um dicionário com essas informações e depois encontre e imprima o nome e a altura da pessoa mais alta e da mais baixa.

```python
pessoas = {}
quantidade = int(input("Quantas pessoas você quer cadastrar? "))

for i in range(quantidade):
    nome = input(f"Digite o nome da {i+1}ª pessoa: ")
    altura = float(input(f"Digite a altura de {nome} (em metros): "))
    pessoas[nome] = altura

# Inicializa com o primeiro item do dicionário
nomes = list(pessoas.keys())
mais_alto = nomes[0]
mais_baixo = nomes[0]

for nome in pessoas:
    if pessoas[nome] > pessoas[mais_alto]:
        mais_alto = nome
    if pessoas[nome] < pessoas[mais_baixo]:
        mais_baixo = nome

print("\n Análise de Alturas:")
print(f"Pessoa mais alta: {mais_alto} com {pessoas[mais_alto]:.2f}m")
print(f"Pessoa mais baixa: {mais_baixo} com {pessoas[mais_baixo]:.2f}m")
```

##5) Simulação de Alfândega com Cadastro Dinâmico de Itens usando dicionário
Simule uma situação de alfândega onde o funcionário cadastra itens novos encontrados na mala de uma pessoa que veio de uma viagem internacional. Após o cadastro de todos os itens, o programa imprime uma nota com todos os itens e seus preços, calcula o valor total e, se o valor total ultrapassar 500 dólares, calcula o valor de 60% a ser pago.


```python
itens = {}
total = 0

print(" Cadastro de itens da mala (digite 'fim' para encerrar)\n")

while True:
    nome_item = input("Nome do item: ")
    if nome_item.lower() == "fim":
        break

    preco = float(input(f"Preço em dólares do item '{nome_item}': "))
    itens[nome_item] = preco
    total += preco

# Impressão da nota
print("\n🧾 Nota Fiscal da Alfândega")
for item, valor in itens.items():
    print(f"- {item}: ${valor:.2f}")

print(f"\n Valor total dos itens: ${total:.2f}")

# Verificação de taxa
if total > 500:
    taxa = total * 0.60
    print(f" Valor excedente! Taxa de 60% aplicada: ${taxa:.2f}")
else:
    print(" Valor dentro do limite. Nenhuma taxa será aplicada.")
```
